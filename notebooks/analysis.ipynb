```python
import sys
sys.path.append("../src")

from data_cleaning import load_trader_data, load_sentiment_data
from feature_engineering import merge_data, create_features
from visualization import save_pnl_by_sentiment, save_winrate_by_sentiment, save_volume_by_sentiment

import pandas as pd
```

```python
trader_df = load_trader_data("../data/raw/historical_data.csv")
sentiment_df = load_sentiment_data("../data/raw/fear_greed_index.csv")

trader_df.head()
```

```python
merged_df = merge_data(trader_df, sentiment_df)
merged_df = create_features(merged_df)

merged_df.head()
```

```python
sentiment_summary = merged_df.groupby("market_sentiment").agg(
    total_trades=("account", "count"),
    total_volume=("usd_value", "sum"),
    avg_pnl=("closed_pnl", "mean"),
    total_pnl=("closed_pnl", "sum"),
    win_rate=("is_profitable", "mean"),
    avg_fee=("fee", "mean")
).reset_index()

sentiment_summary
```

```python
side_summary = merged_df.groupby(
    ["market_sentiment", "side"]
).agg(
    total_trades=("account", "count"),
    avg_pnl=("closed_pnl", "mean"),
    total_pnl=("closed_pnl", "sum"),
    win_rate=("is_profitable", "mean")
).reset_index()

side_summary
```

```python
top_traders = merged_df.groupby("account").agg(
    total_trades=("account", "count"),
    total_pnl=("closed_pnl", "sum"),
    avg_pnl=("closed_pnl", "mean"),
    win_rate=("is_profitable", "mean"),
    total_volume=("usd_value", "sum")
).reset_index()

top_traders = top_traders.sort_values(
    "total_pnl",
    ascending=False
)

top_traders.head(10)
```

```python
top_traders.tail(10)
```

```python
coin_summary = merged_df.groupby("coin").agg(
    total_trades=("coin", "count"),
    total_volume=("usd_value", "sum"),
    total_pnl=("closed_pnl", "sum")
).reset_index()

coin_summary.sort_values(
    "total_volume",
    ascending=False
).head(10)
```

```python
save_pnl_by_sentiment(
    merged_df,
    "../visuals/pnl_by_sentiment.png"
)

save_winrate_by_sentiment(
    merged_df,
    "../visuals/winrate_by_sentiment.png"
)

save_volume_by_sentiment(
    merged_df,
    "../visuals/volume_by_sentiment.png"
)

print("Charts saved successfully")
```
